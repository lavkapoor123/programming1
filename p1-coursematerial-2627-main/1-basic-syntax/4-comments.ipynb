# Comments

Execute the following code cell:

```python
# ask the user for their name
name = input("What's your name?")
# print a greeting for the user
print("Hello " + name)
```

As you can see, the lines starting with hashtags `#` have no impact on the code. These lines are called **comments**. We use them if we simply want to write some text in our code file, often to explain what our code is for. Having the line start with `#` tells the Python interpreter that it simply needs to ignore these lines. If you forget to use the hashtag, the Python interpreter will try to interpret whatever your wrote as code, and this will result in an error:
